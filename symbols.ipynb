# Symbols

```python
from sympy import *
```

```python
x, y, z, a, b = symbols('x y z a b')
```

```python
x = 'abc'
expr = x + 'def'
expr
```

```python
# Substitute
x = symbols('x')
expr = x + 1
expr.subs(x, 2)  # x subs to 2
```

```python
# x + 1 = 4 coding as Eq(x+1, 4) in sympy
Eq(x+1, 4)
```

```python
# (x+1)**2 == x**2+2*x+1 (will putput False) because they are not "Exact" the same
# we use the method a - b = 0 to test if it  is equal
a = (x+1)**2
b = x**2+2*x+1
simplify(a-b)
```

```python
# there is a method called 'equals'
a = cos(x)**2 - sin(x)**2
b = cos(2*x)
a.equals(b)
```

```python
# equation subtraction
c = x**2 - 2*x+1
simplify(a-c)
```

```python
# output fraction in Sympy, use Rational
Rational(1, 2)
```
