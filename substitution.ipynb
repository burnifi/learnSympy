```python
# Basic Operations
from sympy import *
x, y, z = symbols('x y z')
```

```python
# Substitution
expr = cos(x)+1
expr.subs(x, y)  # change x to y
```

```python
# evaluate cos(x) + 1 at the point x = 0
# so we get cos(0)+1, which is 2 because cos(0) = 1
expr.subs(x, 0)
```

```python
# replacing a subexpression with another subexpression
expr = x**y
expr
```

```python
expr = expr.subs(y, x**y)  # change y to x**y
expr
```

```python
expr = sin(2*x) + cos(2*x)
# use expand_trig will expand both
expand_trig(expr)
```

```python
# if we only need to expand the first part, we use subs
expr.subs(sin(2*x), 2*sin(x)*cos(x))
```

```python
expr = cos(x)
expr.subs(x, 0)
```

```python
expr  # original not change
```

```python
# to perform substitutes at once
expr = x**3 +4*x*y -z
expr.subs([(x, 2), (y, 4), (z, 0)])  # use [] to conclude 3 ()
```

```python
# change evwen exponent of x to y
expr = x**4 - 4*x**3 + 4*x**2 - 2*x +3
replacements = [(x**i, y**i) for i in range(5) if i%2 == 0]
expr.subs(replacements)
```

Converting String to Sympy Expression

```python
str_expr = "x**2 + 3*x - 1/2"
expr = sympify(str_expr)
expr
```

```python
expr.subs(x, 2)
```

```python
# to evaluate a numerical expression into a floating point num, use evalf()
expr = sqrt(8)
expr.evalf()
```

```python
pi.evalf(100)
```

```python
expr = cos(2*x)
expr.evalf(subs={x: 2.4})
```

```python
# sometimes there are roundoff errors
one = cos(1)**2 + sin(1)**2
(one - 1).evalf()
```

```python
# above num can be removed by setting the chop flag to True inside evalf()
(one - 1).evalf(chop=True)
```

Lambdify()

```python
import numpy
a = numpy.arange(10)
expr = sin(x)
f = lambdify(x, expr, 'numpy')
f(a)
```

```python
# or use another libraries, e.g. 'math'
f = lambdify(x, expr, 'math')
f(0.1)
```

```python
# with numerical libraries that it does not know about,
# pass a dictionary of  sympy_name: numerical_func pairs. Example
def mysin(x):
  """
  My sine. Note that this is only accurate for small x.
  """
  return x

f = lambdify(x, expr, {'sin': mysin})
f(0.5)
```
