Basic Operations

Shape

- to get the shape of a matrix, use shape() function

```python
from sympy import shape
M = Matrix([[1, 2, 3], [-2, 0, 4]])
M
```

```python
shape(M)
```

Accessing Rows and Columns

- To get an individual row or column of a matrix, use row / col

```python
M.row(0)
```

```python
M.col(-1)
```

Deleting and Inserting Rows and Columns

- to delete a row or a column, use row del or col_del

```python
M.col_del(0)
M
```

```python
M.row_del(1)
M
```

- to insert rows and columns, use row_insert or col_insert

- This operations do not operate in place

```python
M
```

```python
M = M.row_insert(1, Matrix([[0, 4]]))  # insert to [1]
```

```python
M
```

```python
M = M.col_insert(0, Matrix([1, -2]))  # add column in the front
M
```

Advanced Methods

- Determinant -> to compute the determinant of a matrix, use det()

```python
M = Matrix([[1, 0, 1], [2, -1, 3], [4, 3, 2]])
M
```

```python
M.det()
```

RREF - to put a matrix into reduced row echelon form, use rref.

- rref returns a tuple of two elements.

- the first is the reduced row echelon form

- the second is a tuple of indices of the pivot columns

```python
M = Matrix([[1, 0, 1, 3], [2, 3, 4, 7], [-1, -3, -3, -4]])
M
```

```python
M.rref()
```

Nullspace - to find the nullspace of a matrix, use nullspace()

- return a list of column vectors that span the nullspace of the matrix

```python
M = Matrix([[1, 2, 3, 0, 0], [4, 10, 0, 0, 1]])
M
```
