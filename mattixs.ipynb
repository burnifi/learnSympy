Eigenvalues, Eigenvectors, and Diagonalization

- to find the eigenvalues of a matrix, use eigenvals()

- return a dictionary of eignevalue: algebraic_multiplicity pairs (similar to the output of roots)

```python
M = Matrix([[3, -2, 4, -2], [5, 3, -3, -2], [5, -2, 2, -2], [5, -2, -3, 3]])
M
```

```python
M.eigenvals()
```

above means that M has eigenvalues -2, 3, and 5, and that 

- the eigenvalues -2 and 3 have algebraic multiplicity 1

- and that the eigenvalues 5 have algebraic multiplicity 2

to find the eigenvectors of a matrix, use eigenvects().

- eigenvects returns a list of tuples of the form (eigenvalue, algebraic_multiplicity, [eignevectors])

```python
M.eigenvects()
```

-above show us thatthe eigenvalue 5 also has geometric multiplicity 2, because it has two eigenvectors

- Because the algebraic and geometric multiplicities are the same for all the eigenvalues, M is diagonalizable.

```python
# to diagonalize a matrix, use diagonalize()
# reutrn a tuple(P, D), where D is diagonal and M = PDP**-1
P.D = M.diagonalize()
```

Zero testing - if your matrix operations are failing or returning wrong answers

```python
from sympy import *
q = Symbol("q", positive=True)
m = Matrix([
    [-2*cosh(q/3), exp(-q), 1],
    [exp(q), -2*cosh(q/3), 1],
    [1, 1, -2*cosh(q/3)]
])
m.nullspace()
```

```python
import warnings

def my_iszero(x):
    result = x.is_zero

    # Warning if evaluated into None
    if result is None:
        warnings.warn("Zero testing of {} evaluated into None". format(x))
    return result

m.nullspace(iszerofunc=my_iszero)
```

```python
def my_iszero(x):
    result = x.rewrite(exp).simplify().is_zero

    # Warning if evaluated into None
    if result is None:
        warnings.warn("Zero testing of {} evaluated into None".format(x))
    return result

m.nullspace(iszerofunc=my_iszero)
```
