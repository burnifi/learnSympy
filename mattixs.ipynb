To make a matrix in Sympy, use the Matrix()

```python
Matrix([[1, -1], [3, 4], [0, 2]])
```

```python
# to make it easy to make column vectros,a list of elements is considered to be a column vectro
Matrix([1, 2, 3])
```

```python
# Matrices are manupulated just like any other object in Sympy or Pythos
M = Matrix([[1, 2, 3], [3, 2, 1]])
N = Matrix([0, 1, 1])
M*N
```

Basic Operations

Shape

- to get the shape of a matrix, use shape() function

```python
from sympy import shape
M = Matrix([[1, 2, 3], [-2, 0, 4]])
M
```

```python
shape(M)
```

Accessing Rows and Columns

- To get an individual row or column of a matrix, use row / col

```python
M.row(0)
```

```python
M.col(-1)
```

Deleting and Inserting Rows and Columns

- to delete a row or a column, use row del or col_del

```python
M.col_del(0)
M
```

```python
M.row_del(1)
M
```

- to insert rows and columns, use row_insert or col_insert

- This operations do not operate in place

```python
M
```

```python
M = M.row_insert(1, Matrix([[0, 4]]))  # insert to [1]
```

```python
M
```

```python
M = M.col_insert(0, Matrix([1, -2]))  # add column in the front
M
```

Basic Methods

- simple operations sucn as + , * , and **

- to find the inverse of a matrix, just raise it to the -1 power

```python
M = Matrix([[1, 3], [-2, 3]])
N = Matrix([[0, 3], [0, 7]])
M + N
```
