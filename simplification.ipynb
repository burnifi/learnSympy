cancel

cancel() will take any rational function and put it into the dtandard canonical form, p/q
where p and q are expanded polynomials with no common factors,
and the leading coefficients of p and q do not have denominators (i.e. aare integer)

```python
cancel((x**2 + 2*x +1) / (x**2 + x))
```

```python
expr = 1/x + (3*x/2 - 2) / (x - 4)  # run the expression
expr
```

```python
cancel(expr)  # get the final result
```

```python
expr = (x*y**2 - 2*x*y*z + x*z**2 + y**2 - 2*y*z + z**2) / (x**2 -1)
expr
```

```python
cancel(expr)
```

```python
# factor() can also  be used to do the same thing
factor(expr)
```

expand_power_exp / expand_power_base

expand_power_exp() and  expand_power_base() apply identities 1 and 2 from right to left

```python
expand_power_exp(x**(a + b))
```

```python
expand_power_base((x*y)**a)
```

```python
# as with powsimp(), identity 2 is not applied if it is not valid
expand_power_base((z*t)**c)
```

```python
# with powsimp(), you can force the expression to happen without fidding with assumptions by force=True
expand_power_base((z*t)**c, force=True)
```

```python
# with identity 2, identity 1 is applied automatically if the power is a number
# and hence cannot be undone with expand_power_exp()
x**2*x**3
```

```python
expand_power_exp(x**5)
```

powdenest - applies identity 3, from left to right

```python
# as before, the identity is not applied if it is not true under the given assumptions
powdenest((z**a)**b)
```

```python
# And as before, this can be manually overridden with force=True
powdenest((z**a)**b, force=True)
```

```python
# Continued Fractions
def list_to_frac(l):
  expr = Integer(0)
  for i in reversed(l[1:]):
    expr += i
    expr = 1/expr
  return l[0] + expr

list_to_frac([x, y, z])
```

```python
list_to_frac([1,2,3,4])
```

```python
syms = symbols('a0:5')
syms
```

```python
a0, a1, a2, a3, a4 = syms
frac = list_to_frac(syms)
frac
```
